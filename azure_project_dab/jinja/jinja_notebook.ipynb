```python
parameters = [
    {
        "table": "spotify_catalog.silver.factstream",
        "alias": "factstream",
        "cols": "factstream.track_id, factstream.track_name, factstream.artist_id, factstream.album_name"
    },
    {
        "table": "spotify_catalog.silver.dimtrack",
        "alias": "dimtrack",
        "cols": "dimtrack.duration_sec, dimtrack.duration_flag",
        "condition": "dimtrack.track_id = factstream.track_id"
    }
]
```

```python
from jinja2 import Template
```

```python
query_text = """
SELECT
{% for parameter in parameters %}
    {{ parameter.cols }}{% if not loop.last %},{% endif %}
{% endfor %}

FROM {{ parameters[0].table }} AS {{ parameters[0].alias }}

{% for parameter in parameters[1:] %}
LEFT JOIN {{ parameter.table }} AS {{ parameter.alias }}
    ON {{ parameter.condition }}
{% endfor %}
"""
```

```python
jinja_sql_str=Template(query_text)
query=jinja_sql_str.render(parameters=parameters)
print(query)
```

```python
df = spark.sql(query)

display(df)
```
